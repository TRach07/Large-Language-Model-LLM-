# Self-Attention Mechanism Implementation

## Introduction
This notebook implements the self-attention mechanism used in Transformer models. We'll process the sentence "Le chat noir dort." through the complete self-attention pipeline.

## Imports and configuration

```python
import numpy as np

# Dimension configuration
d_embedding = 4
d_head = 2
np.random.seed(42)  # For reproducibility

print("Self-Attention Mechanism Implementation")
print("======================================")
print(f"Embedding dimension: {d_embedding}")
print(f"Head dimension: {d_head}")
print("Sentence: 'Le chat noir dort.'")
```

## Create word embeddings

```python
words = ["Le", "chat", "noir", "dort"]
n_words = len(words)

# Create random embeddings for each word
embeddings = np.random.randn(n_words, d_embedding)

print("Step 1: Word Embeddings")
print("======================")
print("Shape:", embeddings.shape)
print("\nWord embeddings:")
for i, word in enumerate(words):
    print(f"{word}: {embeddings[i]}")
```

## Initialize weight matrices WQ, WK, WV

```python
WQ = np.random.randn(d_embedding, d_head)
WK = np.random.randn(d_embedding, d_head)
WV = np.random.randn(d_embedding, d_head)

print("Step 2: Weight Matrices")
print("======================")
print("WQ shape:", WQ.shape)
print("WK shape:", WK.shape)
print("WV shape:", WV.shape)

print("\nWeight matrix WQ:")
print(WQ)
print("\nWeight matrix WK:")
print(WK)
print("\nWeight matrix WV:")
print(WV)
```

## Calculate Query, Key and Value vectors

```python
Q = np.dot(embeddings, WQ)
K = np.dot(embeddings, WK)
V = np.dot(embeddings, WV)

print("Step 3: Query, Key, Value Vectors")
print("================================")
print("Q shape:", Q.shape)
print("K shape:", K.shape)
print("V shape:", V.shape)

print("\nQuery vectors (Q):")
for i, word in enumerate(words):
    print(f"Q({word}): {Q[i]}")

print("\nKey vectors (K):")
for i, word in enumerate(words):
    print(f"K({word}): {K[i]}")

print("\nValue vectors (V):")
for i, word in enumerate(words):
    print(f"V({word}): {V[i]}")
```

## Calculate attention scores

```python
scores = np.dot(Q, K.T)

print("Step 4: Attention Scores")
print("=======================")
print("Scores shape:", scores.shape)
print("\nRaw attention scores matrix:")
print(scores)

print("\nDetailed attention scores:")
for i in range(n_words):
    for j in range(n_words):
        print(f"Score(Q({words[i]}), K({words[j]})) = {scores[i, j]:.4f}")
```

## Apply softmax

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

attention_weights = softmax(scores)

print("Step 5: Softmax Application")
print("===========================")
print("Attention weights shape:", attention_weights.shape)
print("\nAttention weights (after softmax):")
print(attention_weights)

print("\nVerification (each row sums to 1):")
for i in range(n_words):
    row_sum = np.sum(attention_weights[i])
    print(f"Row {i} ({words[i]}): sum = {row_sum:.6f}")
```

## Calculate weighted output

```python
# Output = attention_weights * V
output = np.dot(attention_weights, V)

print("Step 6: Weighted Output")
print("======================")
print("Output shape:", output.shape)
print("\nFinal weighted vectors:")
for i, word in enumerate(words):
    print(f"Output({word}): {output[i]}")
```

## Final summary and display

```python
print("FINAL RESULTS SUMMARY")
print("====================")

print("\n1. ATTENTION WEIGHTS MATRIX")
print("   (Each row i shows word i's attention to all words j)")
print("\n" + " " * 8 + " ".join(f"{word:>10}" for word in words))
for i, word in enumerate(words):
    print(f"{word:>8}", end="")
    for j in range(n_words):
        print(f"{attention_weights[i, j]:>10.4f}", end="")
    print()

print(f"\n2. FINAL WEIGHTED VECTORS (dimension {d_head})")
for i, word in enumerate(words):
    print(f"   {word}: {output[i]}")

print(f"\n3. COMPARISON WITH INITIAL VALUE VECTORS")
print("   Initial Value vectors:")
for i, word in enumerate(words):
    print(f"   {word}: {V[i]}")
```
